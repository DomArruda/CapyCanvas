```python
import os
import polars as pl

if not os.path.exists("data.csv"):
    partners = [
        "Bob's Tacos",
        "Bullseye",
        "Wallsmart",
        "GreenLeaf Market",
        "Summit Tools",
        "Harbor Coffee",
        "Northwind Outfitters",
        "Cedar Dental",
        "Pixel Pets",
        "Lakeside Hardware",
    ]
    branches = ["Downtown", "Westside", "Airport", "Mall", "Suburban"]

    rows = []
    rid = 1
    for partner in partners:
        for branch in branches:
            rows.append(
                {
                    "Id": rid,
                    "Partner": partner,
                    "Branch": branch,
                    "Rev": 8_000 + (rid * 137) % 25_000,  # varied but deterministic
                }
            )
            rid += 1

    test_data = pl.DataFrame(rows)  # 10 partners × 5 branches = 50 rows
    test_data.write_csv("data.csv")  # polars has no index= kwarg

df = pl.read_csv("data.csv")
df
```

```python
import capycanvas as cc
import polars as pl

template = cc.read_template("app.canvas.html")  # your finished app export
df = pl.read_csv("data.csv") # works for pandas too.


for partner in df["Partner"].unique().to_list():
    filtered = df.filter(pl.col("Partner") == partner)
    out = template.copy(kind="client")   # editor hidden
    out.embed(filtered, name="data.csv") # also accepts path / bytes / pandas
    out.save(f"reports/{partner}.html")
```
